```python
FIGSIZE = (7, 5)

# Separate numeric and non-numeric values
numeric_lifetimes = pd.to_numeric(df["counted_as"], errors="coerce")
df["numeric_lifetime"] = numeric_lifetimes

# Count unspecified and duplicated
n_unspecified = (df["counted_as"] == "unspecified").sum()
n_duplicated = (df["counted_as"] == "duplicated").sum()
n_specified = df["numeric_lifetime"].notna().sum()
n_total = len(df)

# Extract numeric values for plotting
lifetimes = df["numeric_lifetime"].dropna().values

# Calculate descriptive statistics
mean_lifetime = np.mean(lifetimes)
median_lifetime = np.median(lifetimes)

# Mode (from scipy stats)
mode_result = stats.mode(lifetimes, keepdims=True)
mode_lifetime = mode_result.mode[0]
q25, q75 = np.percentile(lifetimes, [25, 75])
iqr = q75 - q25

print(f"Total studies: {n_total}")
print(f"Specified: {n_specified} ({n_specified / n_total * 100:.1f}%)")
print(f"Unspecified: {n_unspecified} ({n_unspecified / n_total * 100:.1f}%)")
print(f"Duplicated: {n_duplicated} ({n_duplicated / n_total * 100:.1f}%)")
print(f"\nDescriptive statistics (specified values only):")
print(f"Mean: {mean_lifetime:.1f} years")
print(f"Median: {median_lifetime:.1f} years")
print(f"Mode: {mode_lifetime:.1f} years")
print(f"IQR: {q25:.1f}–{q75:.1f} years ({iqr:.1f})")
print(f"Range: {lifetimes.min():.1f}–{lifetimes.max():.1f} years")
```

```python
# Figure
fig, ax = plt.subplots(figsize=(4.5, 6), dpi=300)

# Style
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = ["Arial", "Helvetica"]
plt.rcParams["font.size"] = 10
plt.rcParams["axes.linewidth"] = 0.8

# Sort lifetimes
lifetimes_sorted = np.sort(lifetimes)
x_positions = np.arange(len(lifetimes_sorted))

# Main plot
ax.scatter(
    x_positions,
    lifetimes_sorted,
    color="#E69F00",
    marker="o",
    s=30,
    alpha=0.7,
    edgecolors="black",
    linewidths=0.5,
    zorder=3,
)

# Add horizontal reference lines for key statistics
ax.axhline(
    y=median_lifetime,
    color="#0072B2",
    linestyle="-",
    linewidth=1.5,
    label=f"Median: {median_lifetime:.0f} years",
    zorder=2,
)
ax.axhline(
    y=q25,
    color="#0072B2",
    linestyle="--",
    linewidth=1.0,
    alpha=0.6,
    label=f"Q1: {q25:.0f} years",
    zorder=2,
)
ax.axhline(
    y=q75,
    color="#0072B2",
    linestyle="--",
    linewidth=1.0,
    alpha=0.6,
    label=f"Q3: {q75:.0f} years",
    zorder=2,
)

# Create half-violin plot on the right side
# Use KDE to estimate density
kde = stats.gaussian_kde(lifetimes)
y_range = np.linspace(lifetimes.min() - 5, lifetimes.max() + 5, 200)
density = kde(y_range)

# Normalize density to fit in the plot (scale to ~20% of x-axis width)
density_scaled = density / density.max() * len(lifetimes) * 0.25
x_violin = len(lifetimes) + density_scaled

# Plot half-violin
ax.fill_betweenx(
    y_range, len(lifetimes), x_violin, color="#E69F00", alpha=0.3, zorder=1
)
ax.plot(x_violin, y_range, color="#E69F00", linewidth=1.5, zorder=2)

# Add vertical line at base of violin
ax.axvline(
    x=len(lifetimes), color="black", linewidth=0.8, linestyle="-", alpha=0.5, zorder=1
)

# Labels and formatting
ax.set_ylabel("Assumed building lifetime (years)", fontsize=11, fontweight="normal")
ax.set_xlabel(
    "Ranked studies (n = {})".format(n_specified), fontsize=11, fontweight="normal"
)
ax.set_xlim(-2, len(lifetimes) + len(lifetimes) * 0.3)
ax.set_ylim(20, 130)

# Grid
ax.grid(axis="y", linestyle="--", alpha=0.3, linewidth=0.5, zorder=0)
ax.set_axisbelow(True)

# Add legend
ax.legend(loc="upper left", frameon=True, fontsize=9, edgecolor="black", framealpha=0.9)

# Add text box with statistics
stats_text = (
    f"n = {n_specified}\n"
    f"Mean: {mean_lifetime:.0f} years\n"
    f"Median: {median_lifetime:.0f} years\n"
    f"Mode: {mode_lifetime:.0f} years\n"
    f"IQR: {iqr:.0f} years"
)
ax.text(
    0.03,
    0.98,
    stats_text,
    transform=ax.transAxes,
    fontsize=9,
    verticalalignment="top",
    horizontalalignment="left",
    bbox=dict(
        boxstyle="round", facecolor="white", edgecolor="black", alpha=0.9, pad=0.5
    ),
)

# Remove top and right spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

h, l = ax.get_legend_handles_labels()

ax.legend(
    h,
    l,
    ncols=3,
    bbox_to_anchor=(0.5, -0.10),
    loc="upper center",
    frameon=False,
    fontsize=9,
    edgecolor="black",
    framealpha=0.5,
)
# Tighten layout
plt.tight_layout(pad=0.5, rect=[0, 0.05, 1, 1])  # rect reserves space at bottom

# Save figure in high resolution
# file_path = os.path.join(FIG_DIR, 'Figure_4.png')
# plt.savefig(file_path, dpi=1000, bbox_inches='tight')

plt.show()
```
