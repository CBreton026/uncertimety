```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# -----------------------------------------------------------------------------
# Style
# -----------------------------------------------------------------------------
sns.set_theme(style="white", context="paper")

plt.rcParams.update(
    {
        "font.family": "Helvetica",
        "font.size": 22,
        "xtick.labelsize": 22,
        "axes.linewidth": 1.0,
        "xtick.major.width": 1.0,
        "ytick.major.width": 1.0,
        "figure.dpi": 300,
        "savefig.dpi": 600,
    }
)

# -----------------------------------------------------------------------------
# Distribution parameters
# -----------------------------------------------------------------------------
mu1, sigma1 = 60, 18
mu2, sigma2 = 150, 45

dist1 = norm(loc=mu1, scale=sigma1)
dist2 = norm(loc=mu2, scale=sigma2)

# x-range covering ±4σ of both distributions
xmin = min(dist1.ppf(0.0001), dist2.ppf(0.0001))
xmax = max(dist1.ppf(0.9999), dist2.ppf(0.9999))
x = np.linspace(xmin, xmax, 1000)

# -----------------------------------------------------------------------------
# Figure
# -----------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(4.5, 4.5))

ax.plot(
    x,
    dist1.pdf(x),
    color="black",
    linewidth=1.8,
    linestyle="-",
)

ax.plot(
    x,
    dist2.pdf(x),
    color="black",
    linewidth=1.8,
    linestyle="--",
)

# # Mean labels
# y_offset = -0.001
# ax.text(mu1, y_offset, f"{mu1}", ha="center", va="top")
# ax.text(mu2, y_offset, f"{mu2}", ha="center", va="top")

# Formatting
ax.set_xlim(xmin, xmax)
ax.set_ylim(0, max(dist1.pdf(x).max(), dist2.pdf(x).max()) * 1.08)

# Only show ticks at the means
ax.set_xticks([mu1, mu2])
ax.set_xticklabels([f"{mu1}", f"{mu2}"])

# Tick appearance
ax.tick_params(
    axis="x",
    which="major",
    direction="out",
    length=6,
    width=1.0,
    pad=10,  # distance between axis and labels
    bottom=True,
    top=False,
)

# Remove y-axis
ax.set_yticks([])

ax.set_xlabel("")
ax.set_ylabel("")

sns.despine(ax=ax, top=True, right=True, left=True, trim=False)

plt.tight_layout()

plt.savefig("normal_distributions_bw.pdf", bbox_inches="tight")
plt.savefig("normal_distributions_bw.png", dpi=600, bbox_inches="tight")

plt.show()
```
